```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification, DistilBertTokenizer, DistilBertForSequenceClassification
model_name1="distilbert-base-uncased-finetuned-sst-2-english"
mytokenizer1=DistilBertTokenizer.from_pretrained(model_name1)
model1=DistilBertForSequenceClassification.from_pretrained(model_name1)

classifier=pipeline("sentiment-analysis")
res=classifier("I was not happy with class today")
print(res)

model_name2="distilbert-base-uncased-finetuned-sst-2-english"
mymodel2=AutoModelForSequenceClassification.from_pretrained(model_name2)
mytokenizer2=AutoTokenizer.from_pretrained(model_name1)

classifier=pipeline("sentiment-analysis", model=mymodel2, tokenizer=mytokenizer2)
res=classifier("I was not happy with class today")
print(res)
```

```python
from transformers import AutoTokenizer
tokenizer=AutoTokenizer.from_pretrained("bert-base-uncased")
text="I love my Country"

tokens=tokenizer.tokenize(text)
print(tokens)

input_ids=tokenizer.convert_tokens_to_ids(tokens)
print(input_ids)

encoded_input=tokenizer.encode(text)
print(encoded_input)

decoded_input=tokenizer.decode(encoded_input)
print(decoded_input)
```
